```python
import numpy as np
```

```python
# Student Marks Analysis 
# Here is numpy array containing marks of 40 students in eight subjects.
# rows represent 40 students and column represent 8 subjects
np.random.seed(42)
student_marks = np.random.randint(50,100, size = (40,8))
```

```python
student_marks
```

```python
# Checking the dataset 
print("Shape: ", student_marks.shape)
print("Size: ",student_marks.size)
print("Dimensions:", student_marks.ndim)
```

```python
# Overall Statistics
print("Mean = ", np.mean(student_marks))
print("Median = ", np.median(student_marks))
print("Standard Deviation = ", np.std(student_marks))
print("Minimum = ", np.min(student_marks))
print("Maximum = ", np.max(student_marks))
print("Total Marks = ", np.sum(student_marks))
```

```python
# Average marks of each student
avg_marks_student = np.mean(student_marks, axis = 1)
avg_marks_student
```

```python
# Average marks in each subject
avg_marks_subject = np.mean(student_marks, axis = 0)
print(avg_marks_subject)
```

```python
# Find the highest-scoring student
best_student_index = np.argmax(np.max(student_marks, axis = 1))
print("Best Student = ", best_student_index + 1)
print("Highest Score made by best student = ", np.max(student_marks[best_student_index]))
```

```python
# Find the lowest-scoring student
worst_stu_index = np.argmin(np.min(student_marks, axis = 1))
print("Worst Student= ", worst_stu_index + 1)
print("Lowest score made by the worst student = ", np.min(student_marks[worst_stu_index]))
```

```python
# Student who score above 95 in any subject
mask = np.any(student_marks > 95, axis = 1)
above_95_scorer = student_marks[mask]
print(above_95_scorer)
```

```python
# Top 10% scores
threshold = np.percentile(student_marks, 90)
print("Top 10% threshold:", threshold)
mask = student_marks >= threshold
print("10% score = ",student_marks[mask])
```
