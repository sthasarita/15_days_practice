```python
# filling the missing values
# Replace hidden missing-value strings with actual NaN
df['Final_Exam_Score'] = df['Final_Exam_Score'].replace(['missing', 'N/A', ' ', '-'], np.nan)

# Now convert the column to numeric (it's still stored as text/object)
df['Final_Exam_Score'] = pd.to_numeric(df['Final_Exam_Score'])

df['Final_Exam_Score'].isnull().sum()
```

```python
# Attendance can't exceed 100%
df.loc[df['Attendance_Rate'] > 100, 'Attendance_Rate'] = np.nan

# Scores can't be negative
df.loc[df['Final_Exam_Score'] < 0, 'Final_Exam_Score'] = np.nan

# Midterm score of 999 is clearly a data-entry error (max possible should be 100)
df.loc[df['Midterm_Score'] > 100, 'Midterm_Score'] = np.nan
```

```python
df['Tuition_Paid_USD'] = df['Tuition_Paid_USD'].replace('PENDING', np.nan)
df['Tuition_Paid_USD'] = df['Tuition_Paid_USD'].str.replace('$', '', regex=False)
df['Tuition_Paid_USD'] = df['Tuition_Paid_USD'].str.replace(',', '', regex=False)
df['Tuition_Paid_USD'] = df['Tuition_Paid_USD'].str.strip()

df['Tuition_Paid_USD'] = pd.to_numeric(df['Tuition_Paid_USD'])

df['Tuition_Paid_USD'].unique()
```

```python
# Count missing values per row
df['missing_count'] = df.isnull().sum(axis=1)

# Sort so the most complete row (fewest missing values) comes first within each Student_ID group
df = df.sort_values(['Student_ID', 'missing_count'])

# Keep only the first row per Student_ID (i.e. the most complete one)
df = df.drop_duplicates(subset=['Student_ID'], keep='first')

# Drop the helper column, we don't need it anymore
df = df.drop(columns=['missing_count'])

df.shape
```

 ### Grouping

```python
# Average scores grouped by whether the student passed the prep course
df.groupby('Passed_Prep_Course')[['Midterm_Score', 'Final_Exam_Score', 'Attendance_Rate']].mean()
```

```python
# Average scores grouped by payment status
df.groupby('Payment_Status')[['Midterm_Score', 'Final_Exam_Score', 'Attendance_Rate']].mean()
```

```python
# Count of students in each group
df.groupby('Passed_Prep_Course').size()
```

```python
# Multiple aggregations at once using .agg()
df.groupby('Passed_Prep_Course').agg(
    avg_final=('Final_Exam_Score', 'mean'),
    max_final=('Final_Exam_Score', 'max'),
    min_final=('Final_Exam_Score', 'min'),
    avg_attendance=('Attendance_Rate', 'mean'),
    student_count=('Student_ID', 'count')
).round(2)
```

```python
# Grouping by two columns at once (prep course AND payment status)
df.groupby(['Passed_Prep_Course', 'Payment_Status'])['Final_Exam_Score'].mean().round(2)
```

### Statistical analysis


```python
# Correlation between the numeric columns:
df[['Attendance_Rate', 'Midterm_Score', 'Final_Exam_Score']].corr()
```

```python
# Add a derived column: overall average score, and a Pass/Fail flag
df['Overall_Avg'] = df[['Midterm_Score', 'Final_Exam_Score']].mean(axis=1).round(2)
df['Result'] = np.where(df['Overall_Avg'] >= 50, 'Pass', 'Fail')
df['Result'].value_counts()
```
