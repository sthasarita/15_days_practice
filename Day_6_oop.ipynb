```python
# OOP: Bank Management System
class Customer:
    bank_name = "Nepal Bank Limited"

    def __init__(self, customer_id, name, phone):
        self.customer_id = customer_id
        self.name = name
        self.phone = phone
    def display_customer(self):
        print(f"Customer ID: {self.customer_id}")
        print(f"Name: {self.name}")
        print(f"Phone: {self.phone}")

    @classmethod
    def change_bank_name(cls, new_name):
        cls.bank_name = new_name

    @staticmethod 
    def validate_phone(phone):
        return phone.isdigit() and len(phone) == 10

class Account:
    def __init__(self, acc_num, cus, init_balance = 0):
        self.acc_num = acc_num
        self.cus = cus
        # Encapsulating the balance and transactions attributes
        self.__balance = init_balance
        self.__transactions = []

        if init_balance > 0:
            self.__transactions.append(
                f"Initial deposit: Rs. {init_balance}"
            )
        #Instance method
    def deposit(self, amount):
        self.__balance += amount
        self.__transactions.append(f"Deposited: Rs. {amount}")
        print(f"Rs. {amount} deposited successfully.")

    #Instance method
    def withdraw(self, amount):
        if self.__balance <=0:
            print("No balance in account")
        else:
            self.__balance -= amount
            self.__transactions.append(f"Withdrawn: Rs. {amount}")
            print(f"Rs. {amount} has been withdrawn.")
    def check_balance(self):
        print(f"Current Balance = Rs. {self.__balance}")
    def show_transactions(self):
        print("All Transactions:")
        for t in self.__transactions:
            print("-", t)
    def display_account(self):
        print("---Account Details---")
        print(f"Account Number: {self.acc_num}")
        print(f"Customer: {self.cus}")
        print(f"Bank: {Customer.bank_name}")
        self.check_balance()

c1 = Customer(
    customer_id = 101,
    name = "Sarita",
    phone = "9777777777"
)

print("Customer Info:")
c1.display_customer()

a1 = Account(
    acc_num="ACC1001",
    cus=c1,
    init_balance=5000
)
a1.display_account()
print("\n--- Banking Operations ---")
a1.deposit(2000)
a1.withdraw(1000)
a1.check_balance()
a1.withdraw(10000)
a1.show_transactions()

Customer.change_bank_name("Global IME Bank")

print("Updated Bank Name:", Customer.bank_name)
```

```python
# Vehicle Management System
from abc import ABC, abstractmethod

class Vehicle(ABC):
    def __init__(self, registration_number, owner_name, color, price):
        self.registration_number = registration_number
        self.owner_name = owner_name
        self.color = color
        self.price = price

    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def vehicle_type(self):
        pass

    def display_info(self):
        print(f"Registration Number: {self.registration_number}")
        print(f"Owner Name: {self.owner_name}")
        print(f"Color: {self.color}")
        print(f"Price: Rs. {self.price}")

    # Dunder method
    def __str__(self):
        return f"{self.vehicle_type()} - {self.registration_number}"

class Petrol:
    def refuel(self):
        print("Vehicle is being refueled.")

class Electric:
    def charge(self):
        print("Vehicle is charging.")

class Car(Vehicle):
    def start(self):
        print("Car starts with a key.")
    # Method overriding
    def vehicle_type(self):
        return "Car"

class Bike(Vehicle):
    def start(self):
        print("Bike starts with self-start or kick.")
    # Method overriding
    def vehicle_type(self):
        return "Bike"
        
# Multiple Inheritance
class Truck(Vehicle, Petrol):
    def start(self):
        print("Truck starts with a heavy engine.")
    # Method overriding
    def vehicle_type(self):
        return "Truck"

# Electric Car
# Multiple Inheritance
class ElectricCar(Car, Electric):
    def start(self):
        print("Electric car starts silently.")
    # Method overriding
    def vehicle_type(self):
        return "Electric Car"

# Creating Objects
car = Car("1234","Ram","White",3500000)
bike = Bike("5678","Sita","Black",250000)
truck = Truck("9090","Hari","Blue",8500000)
electric_car = ElectricCar("2020","Gita","Red",5500000)

# Display Information
print("CAR")
car.display_info()

print("BIKE")
bike.display_info()

print("TRUCK")
truck.display_info()

print("ELECTRIC CAR")
electric_car.display_info()

# Polymorphism
print("\nPOLYMORPHISM")
vehicles = [car, bike, truck, electric_car]
for vehicle in vehicles:
    print(f"\n{vehicle}")
    vehicle.start()

# Multiple Inheritance
print("\nMULTIPLE INHERITANCE")
truck.refuel()
electric_car.charge()
```
