```python
# Custom Decorator for Logging
def log_function(func):
    def wrapper(*args, **kwargs):
        print(f"Starting function: {func.__name__}")
        result = func(*args, **kwargs)
        print(f"Finished function: {func.__name__}")
        return result
    return wrapper
    
@log_function
def add(a, b):
    return a + b

result = add(10, 20)
print("Result:", result)
```

```python
# Fibonacci Generator
def fibonacci(n):
    a = 0
    b = 1
    for i in range(n):
        yield a
        a, b = b, a + b
        
for number in fibonacci(10):
    print(number, end=" ")
```

```python
# Even Number Generator
def even_numbers(start, end):
    for number in range(start, end + 1):
        if number % 2 == 0:
            yield number
            
for number in even_numbers(1, 20):
    print(number, end=" ")
```

```python
# Exception-Safe Calculator
def calculator():
    try:
        num1 = float(input("Enter first number: "))
        operator = input("Enter operator (+, -, *, /): ")
        num2 = float(input("Enter second number: "))
        if operator == "+":
            result = num1 + num2
        elif operator == "-":
            result = num1 - num2
        elif operator == "*":
            result = num1 * num2
        elif operator == "/":
            result = num1 / num2
        else:
            print("Invalid operator.")
            return
        print("Result:", result)
    except ValueError:
        print("Please enter valid numbers.")
    except ZeroDivisionError:
        print("Cannot divide by zero.")
        
calculator()
```

```python
# Function Using *args and **kwargs
def student_info(*args, **kwargs):
    print("Subjects:")
    for subject in args:
        print("-", subject)
    print("\nStudent Information:")
    for key, value in kwargs.items():
        print(key, ":", value)

student_info(
    "Math",
    "Science",
    "Computer",
    name="Ram",
    age=20,
    city="Kathmandu"
)
```

```python
# Expense Tracker
from functools import wraps

expenses = []

# Logging Decorator
def log_action(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print(f"\n[LOG] Running: {func.__name__}")
        result = func(*args, **kwargs)
        print(f"[LOG] Finished: {func.__name__}")
        return result
    return wrapper

# Add Expense
@log_action
def add_expense(**kwargs):
    expense = {
        "title": kwargs["title"],
        "amount": kwargs["amount"],
        "category": kwargs["category"]
    }
    expenses.append(expense)
    print("Expense added successfully!")

# Generator
def expense_generator():
    for expense in expenses:
        yield expense

# View Expenses
@log_action
def view_expenses():
    if not expenses:
        print("No expenses available.")
        return
    print("\n--- All Expenses ---")
    for i, expense in enumerate(expense_generator(), start=1):
        print(
            f"{i}. {expense['title']} | "
            f"Rs. {expense['amount']} | "
            f"{expense['category']}"
        )

# Calculate Total
@log_action
def calculate_total():
    if not expenses:
        print("No expenses available.")
        return
    amounts = list(map(lambda expense: expense["amount"], expenses))
    total = sum(amounts)
    print(f"Total Expenses: Rs. {total:.2f}")


# Search by Category
@log_action
def search_by_category(category):
    results = list(
        filter(
            lambda expense: expense["category"].lower() == category.lower(),
            expenses
        )
    )
    if not results:
        print("No expenses found in this category.")
        return
    print(f"\n--- {category} Expenses ---")
    for expense in results:
        print(
            f"{expense['title']} | "
            f"Rs. {expense['amount']}"
        )

# Delete Expense
@log_action
def delete_expense(index):
    if index < 1 or index > len(expenses):
        print("Invalid expense number.")
        return
    removed = expenses.pop(index - 1)
    print(f"Deleted: {removed['title']}")

# Menu
def main():
    while True:
      
        print("---EXPENSE TRACKER---")
        print("1. Add Expense")
        print("2. View Expenses")
        print("3. Calculate Total")
        print("4. Search by Category")
        print("5. Delete Expense")
        print("6. Exit")
        choice = input("Enter your choice: ")
        if choice == "1":
            try:
                title = input("Enter expense title: ")
                amount = float(
                    input("Enter amount: ")
                )
                if amount <= 0:
                    raise ValueError(
                        "Amount must be greater than zero."
                    )
                category = input(
                    "Enter category: "
                )
                add_expense(
                    title=title,
                    amount=amount,
                    category=category
                )
            except ValueError as error:
                print("Error:", error)
        elif choice == "2":
            view_expenses()
        elif choice == "3":
            calculate_total()
        elif choice == "4":
            category = input(
                "Enter category to search: "
            )
            search_by_category(category)
        elif choice == "5":
            view_expenses()
            if expenses:
                try:
                    number = int(
                        input(
                            "Enter expense number to delete: "
                        )
                    )
                    delete_expense(number)
                except ValueError:
                    print(
                        "Please enter a valid number.")
        elif choice == "6":
            print("Thank you for using Expense Tracker!")
            break
        else:
            print("Invalid choice. Please try again.")
# Program starts here
main()
```
